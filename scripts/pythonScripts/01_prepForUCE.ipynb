```python
# Import data
import scanpy as sc
cow = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/cow.h5ad")
dog = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/dog.h5ad")
goat = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/goat.h5ad")
guineaPig = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/guineaPig.h5ad")
human = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/human.h5ad")
macaque = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/macaque.h5ad")
mouse = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/mouse.h5ad")
pig = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/pig.h5ad")
rabbit = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/rabbit.h5ad")
rat = sc.read_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/rat.h5ad")
```

```python
# Tag species so you can track it after embedding
cow.obs["species"] = "cow"
dog.obs["species"] = "dog"
goat.obs["species"] = "goat"
guineaPig.obs["species"] = "guineaPig"
human.obs["species"] = "human"
macaque.obs["species"] = "macaque"
mouse.obs["species"] = "mouse"
pig.obs["species"] = "pig"
rabbit.obs["species"] = "rabbit"
rat.obs["species"] = "rat"
```

```python
# Save prepped datasets 
cow.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/cow_prepped.h5ad")
dog.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/dog_prepped.h5ad")
goat.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/goat_prepped.h5ad")
guineaPig.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/guineaPig_prepped.h5ad")
human.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/human_prepped.h5ad")
macaque.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/macaque_prepped.h5ad")
mouse.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/mouse_prepped.h5ad")
pig.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/pig_prepped.h5ad")
rabbit.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/rabbit_prepped.h5ad")
rat.write_h5ad("/work/hcn4/260630_vertCons_wd/scTrx/annDataObjs/rat_prepped.h5ad")
```
